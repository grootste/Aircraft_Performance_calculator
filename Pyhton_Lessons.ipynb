```python
import math

def pow(a,b):
  return math.pow(a,b)

def ceilling(a):
  return math.ceil(a)

def floor(a):
  return math.floor(a)

def factorial(a):
  return math.factorial(a)


a = 5
b = 2
c = 5.5
print(pow(a,b))
print(ceilling(c))
print(floor(c))
print(factorial(a))
```
